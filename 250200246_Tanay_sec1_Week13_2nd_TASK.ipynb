```python
bikes_data = []

for bike in bikes:

  title = bike.h2.text.strip()

  Rent_full= bike.find('div',class_="bhc-card-price-row").text
  Rent_rate_per_day= Rent_full.split('₹')[1].split('/')[0].strip()


  Engine = bike.find('div',class_="bhc-card-specs").text
  Engine_in_cc= Engine.split('Engine')[1].split('Gearbox')[0].strip()


  Gearbox_type = Engine.split('Gearbox')[1].split('Allowance')[0].strip()


  bikes_data.append({
      "title":title,
      "Rent_rate_per_day": Rent_rate_per_day,
      "Engine_in_cc": Engine_in_cc,
      "Gearbox_type": Gearbox_type
  })

bikes_data
```

```python
import pandas as pd
df=pd.DataFrame(bikes_data)
```
